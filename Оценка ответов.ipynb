Критерии оценки качества ответа на один запрос:
* Пусть N - число построенных по запросу графиков.
* Каждый график отценивается отдельно по следующим критериям:
  - 0 баллов, если код не запускается.
  - 0.5 баллов, если код запускается, но предложенный график некрасиво оформлен или не подходит под описание запроса.
  - 1 балл, если график строится и удовлетворяет всем требованиям.
* Усредненные по всем графикам оценка ставится, как оценка ответа.

Даны массивы Y = [3, 6, 8, 2, 5, 7] и X = [1, 2, 3, 4, 5, 6], построй график y от x

```python
import matplotlib.pyplot as plt
import numpy as np

# Данные
X = np.array([1, 2, 3, 4, 6, 7])
Y = np.array([3, 6, 8, 2, 5, 7])

# Создание фигуры и осей
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# Линейный график
ax[0].plot(X, Y, 
           color='blue',          # Цвет линии
           linewidth=2,           # Толщина линии
           marker='o',            # Маркеры точек
           markersize=8,          # Размер маркеров
           markerfacecolor='red',  # Цвет маркеров
           markeredgecolor='black',# Цвет границ маркеров
           markeredgewidth=1,     # Толщина границ маркеров
           linestyle='-',          # Стиль линии
           alpha=0.8)             # Прозрачность
ax[0].set_xlabel('X', fontsize=12)
ax[0].set_ylabel('Y', fontsize=12)
ax[0].set_title('Линейный график y от x', fontsize=14)
ax[0].grid(True, color='gray', alpha=0.3, linestyle=':')

# График рассеяния
ax[1].scatter(X, Y, 
              color='green',       # Цвет точек
              s=100,              # Размер точек
              alpha=0.7,          # Прозрачность
              edgecolor='black',   # Цвет границ
              linewidths=1)        # Толщина границ
ax[1].set_xlabel('X', fontsize=12)
ax[1].set_ylabel('Y', fontsize=12)
ax[1].set_title('График рассеяния y от x', fontsize=14)
ax[1].grid(True, color='gray', alpha=0.3, linestyle=':')

plt.tight_layout()
plt.show()
```

Имеется pandas.DataFrame data с полями months и revenue, нарисуй гистограмму по этим данным

```python
import pandas as pd

data = {
    'months': ['January', 'February', 'March', 'April', 'May', 'June'],
    'revenue': [15000, 18000, 22000, 19500, 24000, 21000]
}

df = pd.DataFrame(data)
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Создаем гистограмму
plt.hist(df['revenue'], bins=10, 
         color='skyblue',          # Цвет столбцов
         edgecolor='black',        # Цвет границ
         linewidth=1,              # Толщина границ
         alpha=0.7)                # Прозрачность

# Добавляем подписи и заголовок
plt.xlabel('Revenue', fontsize=12, color='navy')
plt.ylabel('Frequency', fontsize=12, color='navy')
plt.title('Histogram of Revenue', fontsize=14, color='darkblue', pad=20)

# Включаем сетку
plt.grid(True, alpha=0.3, linestyle='--')

# Отображаем график
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Группируем данные по месяцам и суммируем revenue
monthly_revenue = df.groupby('months')['revenue'].sum().reset_index()

# Создаем гистограмму
plt.bar(monthly_revenue['months'], monthly_revenue['revenue'], 
        color='skyblue',          # Цвет столбцов
        edgecolor='black',        # Цвет границ
        linewidth=1,              # Толщина границ
        alpha=0.7)                # Прозрачность

# Добавляем подписи и заголовок
plt.xlabel('Months', fontsize=12, color='navy')
plt.ylabel('Revenue', fontsize=12, color='navy')
plt.title('Monthly Revenue', fontsize=14, color='darkblue', pad=20)

# Включаем сетку
plt.grid(True, alpha=0.3, linestyle='--')

# Отображаем график
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Создаем гистограмму с помощью seaborn
sns.histplot(data=df, x='months', y='revenue', 
             color='green',                             # Цвет заливки
             edgecolor='white',                         # Цвет границ
             linewidth=1.5,                             # Толщина границ
             alpha=0.6,                                 # Прозрачность
             stat='identity')                           # Параметр для суммирования значений

# Добавляем подписи и заголовок
plt.xlabel('Months', fontsize=12)
plt.ylabel('Revenue', fontsize=12)
plt.title('Monthly Revenue (seaborn)', fontsize=14)

# Отображаем график
plt.show()
```

Построй круговую диаграмму по массиву months.revenue

```python
import matplotlib.pyplot as plt

# Предположим, что у вас есть массив данных months.revenue
months = ['January', 'February', 'March', 'April']
revenue = [1000, 2000, 1500, 3000]  # Пример данных

# Создаем круговую диаграмму
plt.figure(figsize=(8, 8))
plt.pie(revenue, 
        labels=months, 
        autopct='%1.1f%%', 
        startangle=90, 
        shadow=True, 
        textprops={'fontsize': 12, 'color': 'darkblue'},
        wedgeprops={'edgecolor': 'black', 'linewidth': 2, 'linestyle': '-'})

plt.title('Monthly Revenue Distribution', fontsize=14, pad=20)
plt.axis('equal')
plt.show()
```

Сгенерируй и нарисуй векторное поле

```python
import matplotlib.pyplot as plt
import numpy as np

# Создаем сетку координат
Y, X = np.mgrid[-3:3:100j, -3:3:100j]

# Задаем компоненты вектора поля
U = -1 - X**2 + Y
V = 1 + X - Y**2

# Создаем график
fig, ax = plt.subplots(figsize=(10, 8))

# Вычисляем скорость для окраски линий
speed = np.sqrt(U**2 + V**2)

# Создаем streamplot
strm = ax.streamplot(X, Y, U, V, 
                     density=2,              # Плотность линий потока
                     color=speed,            # Цвет линий зависит от скорости
                     linewidth=1.5,          # Толщина линий
                     cmap='plasma',          # Цветовая карта
                     arrowsize=1.5,          # Размер стрелок
                     arrowstyle='->',        # Стиль стрелок
                     minlength=0.1,          # Минимальная длина линии
                     maxlength=4.0)          # Максимальная длина линии

# Добавляем цветовую шкалу
plt.colorbar(strm.lines, ax=ax, label='Скорость потока', shrink=0.8)

# Настройка графика
ax.set_xlabel('X', fontsize=12)
ax.set_ylabel('Y', fontsize=12)
ax.set_title('Streamplot - Векторное поле потока', fontsize=14)
ax.set_aspect('equal')  # Сохраняем пропорции
ax.grid(True, alpha=0.3)  # Добавляем сетку

plt.show()
```

Нарисуй график y=x^2 и сделай его красным

```python
import matplotlib.pyplot as plt
import numpy as np

# Создаем данные для графика
x = np.linspace(-10, 10, 100)  # Создаем массив x от -10 до 10 из 100 точек
y = x ** 2  # Вычисляем y = x^2

# Рисуем график
plt.plot(x, y, 
         color='red',    # Цвет линии - красный
         linewidth=2,    # Толщина линии
         label='y = x²')  # Подпись для легенды

# Добавляем заголовок и подписи осей
plt.title('График функции y = x²', fontsize=14)
plt.xlabel('x', fontsize=12)
plt.ylabel('y', fontsize=12)

# Добавляем легенду
plt.legend()

# Показываем сетку
plt.grid(True, color='gray', alpha=0.3, linestyle=':')

# Отображаем график
plt.show()
```

Построй в одной сетке графики y = x^2 и y = x^3

```python
import matplotlib.pyplot as plt
import numpy as np

# Создаем данные
x = np.linspace(-2, 2, 100)
y1 = x**2
y2 = x**3

# Создаем график
plt.figure(figsize=(10, 6))

# Рисуем первую линию
plt.plot(x, y1, 
         color='red', 
         linewidth=2, 
         label='y = x²')

# Рисуем вторую линию
plt.plot(x, y2, 
         color='blue', 
         linewidth=2, 
         label='y = x³')

# Добавляем заголовок и метки осей
plt.title('Графики функций y = x² и y = x³', fontsize=14)
plt.xlabel('x', fontsize=12)
plt.ylabel('y', fontsize=12)

# Добавляем легенду
plt.legend(fontsize=12)

# Добавляем сетку
plt.grid(True, color='gray', alpha=0.3, linestyle=':')

# Отображаем график
plt.show()
```
